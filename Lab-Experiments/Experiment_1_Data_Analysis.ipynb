**Experiment-1.1**

**Read the employee dataset and display the average salary of each department using a bar chart.**

```python
import pandas as pd
```

```python
df = pd.read_csv('/content/employee_information_100.csv')
```

```python
df.head()
```

```python
avg_salary_by_dept = df.groupby('Department')['Salary'].mean()
```

```python
print(avg_salary_by_dept)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize = (8, 5))
avg_salary_by_dept.plot(kind = 'bar')
plt.xlabel('Department')
plt.ylabel('Average Salary')
plt.title('Average Salary of Each Department')
plt.show()
```

**Experiment-1.3**


**Display the percentage of male and female employees using a pie chart**

```python
gender_count = df["Gender"].value_counts()
```

```python
print(gender_count)
```

```python
plt.figure(figsize = (6, 6))
gender_count.plot(kind = 'pie', autopct = '%1.1f%%')
plt.axis('equal')
plt.title('Percentage of Male and Female Employees')
plt
```

**Experiment-1.4**

**Draw a histogram showing salary distribution.**

```python
plt.figure(figsize = (8, 5))
plt.hist(df['Salary'], bins = 10, edgecolor = 'k')
plt.xlabel('Salary')
plt.ylabel('Frequency')
plt.title('Salary Distribution')
plt.show()
```

**Experiment-1.5**

**Draw a scatter plot showing the relationship between experience and salary.**

```python
plt.figure(figsize = (8, 5))
plt.scatter(df['Experience_Years'], df['Salary'])
plt.xlabel('Experience')
plt.ylabel('Salary')
plt.title('Experience vs Salary')
plt.show()
```
