# **4.1**

**To perform Term-Frequency Analysis and Named Entity Recognition on a given text using any toolkit.**

**Calculate the frequency of each word and save the output in CSV format (Term, Frequency)
Identify the top 10 most frequent terms and display their corresponding frequencies.**


```python
import re
import csv
from collections import Counter

with open("4.1_4.2_input.txt", "r", encoding="utf-8") as file:
    text = file.read()

words = re.findall(r"[A-Za-z]+", text.lower())

frequency = Counter(words)

with open("term_frequency.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(["Term", "Frequency"])

    for term, freq in frequency.most_common():
        writer.writerow([term, freq])

print("Top 10 Most Frequent Terms:")
for term, freq in frequency.most_common(10):
    print(term, freq)
```

# **4.2**

**To perform Term-Frequency Analysis on a given text without using any toolkit.**

**Calculate the frequency of each word and save the output in CSV format (Term, Frequency)
Identify the top 10 most frequent terms and display their corresponding frequencies.**


```python
import re
import csv

with open("4.1_4.2_input.txt", "r", encoding="utf-8") as file:
    text = file.read()

words = re.findall(r"[A-Za-z]+", text.lower())

frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

sorted_frequency = sorted(
    frequency.items(),
    key=lambda x: x[1],
    reverse=True
)

with open("term_frequency.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(["Term", "Frequency"])

    for term, freq in sorted_frequency:
        writer.writerow([term, freq])

print("Top 10 Most Frequent Terms:")
for term, freq in sorted_frequency[:10]:
    print(term, freq)
```

# **4.3**

**Write a Python program without using NLTK, spaCy, or any other NLP toolkit to:**

**Read multiple text documents.**

**Convert text to lowercase.**

**Remove punctuation.**

**Tokenize the documents.**

**Calculate TF for every term.**

**Calculate DF for every term.**

**Calculate IDF using the logarithmic formula.**

**Calculate TF-IDF for every term in every document.**

**Display Term, TF, DF, IDF, and TF-IDF at each step.**

**Identify the top 10 terms with the highest TF-IDF score for each document.**


```python
import re
import math
import csv

#1.Read multiple text documents.
documents = {
    "Document 1": "Natural language processing is a field of artificial intelligence.",
    "Document 2": "Natural language processing helps computers understand human language.",
    "Document 3": "Machine learning is an important part of artificial intelligence."
}

tokenized_documents = {}

for doc_name, text in documents.items():
    #2. Convert text to lowercase
    text = text.lower()
    #3. Remove punctuation
    text = re.sub(r'[^\w\s]', '', text)
    #4. Tokenize the documents.
    words = text.split()
    tokenized_documents[doc_name] = words

print("TOKENIZED DOCUMENTS")
print("--------------------")

for doc_name, words in tokenized_documents.items():
    print(doc_name, ":", words)


#5. Calculate TF for every term in every document
tf = {}

for doc_name, words in tokenized_documents.items():

    tf[doc_name] = {}

    total_words = len(words)

    for word in words:
        if word in tf[doc_name]:
            tf[doc_name][word] += 1
        else:
            tf[doc_name][word] = 1

    for word in tf[doc_name]:
        tf[doc_name][word] = tf[doc_name][word] / total_words


#6. Calculate DF for every term
df = {}

for doc_name, words in tokenized_documents.items():

    unique_words = set(words)

    for word in unique_words:
        if word in df:
            df[word] += 1
        else:
            df[word] = 1


#7. Calculate IDF
total_documents = len(documents)

idf = {}

for word in df:
    idf[word] = math.log(total_documents / df[word])


#8. Calculate TF-IDF
tf_idf = {}

for doc_name in tokenized_documents:

    tf_idf[doc_name] = {}

    for word in tf[doc_name]:
        tf_idf[doc_name][word] = tf[doc_name][word] * idf[word]


#9. Display Term, TF, DF, IDF and TF-IDF
print("\nTF-IDF RESULTS")
print("--------------")

for doc_name in documents:

    print("\n", doc_name)
    print("Term\t\tTF\t\tDF\t\tIDF\t\tTF-IDF")

    for word in tf[doc_name]:
        print(
            f"{word:<15}"
            f"{tf[doc_name][word]:.4f}\t\t"
            f"{df[word]}\t\t"
            f"{idf[word]:.4f}\t\t"
            f"{tf_idf[doc_name][word]:.4f}"
        )


#10. Identify top 10 terms with highest TF-IDF for each document
print("\nTOP 10 TERMS FOR EACH DOCUMENT")
print("-------------------------------")

for doc_name in tf_idf:

    sorted_terms = sorted(
        tf_idf[doc_name].items(),
        key=lambda x: x[1],
        reverse=True
    )

    print("\n", doc_name)

    for term, score in sorted_terms[:10]:
        print(term, ":", round(score, 4))


# Save TF-IDF results to CSV
with open("tf_idf_results.csv", "w", newline="", encoding="utf-8") as file:

    writer = csv.writer(file)

    writer.writerow([
        "Document",
        "Term",
        "TF",
        "DF",
        "IDF",
        "TF-IDF"
    ])

    for doc_name in tf_idf:

        for word in tf_idf[doc_name]:

            writer.writerow([
                doc_name,
                word,
                round(tf[doc_name][word], 4),
                df[word],
                round(idf[word], 4),
                round(tf_idf[doc_name][word], 4)
            ])

print("\nTF-IDF results saved to tf_idf_results.csv")
```
