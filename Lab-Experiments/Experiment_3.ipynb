# Experiment 3.1:

**Write a Python program to perform stemming using the Porter Stemmer. Apply stemming to the following words. Display the original words and their stems.**

**(playing, played, plays, studies, studying, connected, connection, computers)**


```python
from nltk.stem import PorterStemmer

ps = PorterStemmer()

words = ["playing", "played", "plays", "studies", "studying",
         "connected", "connection", "computers"]

for word in words:
    print(word, "->", ps.stem(word))
```

# Experiment 3.2:  
**Write a Python program to perform lemmatization on a given input words:**

**(cats,	dogs,   running,  runs,	ran, studies, studying,	better, children, mice,	went, ate,
leaves,	caring)**


```python
import nltk
from nltk.stem import WordNetLemmatizer

nltk.download('wordnet')

lemmatizer = WordNetLemmatizer()

words = ["cats", "dogs", "running", "runs", "ran", "studies",
         "studying", "better", "children", "mice", "went", "ate",
         "leaves", "caring"]

for word in words:
    print(word, "->", lemmatizer.lemmatize(word))
```
